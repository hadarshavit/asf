```python
import pandas as pd
import plotly.express as px
import scipy.stats as stats
import numpy as np
```

```python
display_names = {
    "BeamSearch_w10": "Beam Search",
    "BeamSearch_w5": "Beam Search",
    "BruteForce": "Brute Force",
    "GeneticAlgorithm_large": "Genetic Algorithm",
    "GeneticAlgorithm_small": "Genetic Algorithm",
    "MarginalContribution_Backward": "Marginal Contrib. (Backward)",
    "MarginalContribution_Forward": "Marginal Contrib. (Forward)",
    "OptimizeDE": "Optimize DE",
    "RandomLocalSearch_r10": "Random Local Search",
    "RandomLocalSearch_r5": "Random Local Search",
    "SBS": "Single Best Solver",
}
```

```python
df = pd.read_csv("/home/shavit/asf/bench/results/pre_selector_benchmark.csv")
```

```python
df = df.drop(columns="selected_algorithms")
```

```python
for scenario in df["scenario"].unique():
    print(f"Scenario: {scenario}")
    for c in sorted(df["n_algorithms_selected"].unique()):
        print(f"  n_algorithms_selected: {c}")
        cdf = df.loc[
            (df["scenario"] == scenario) & (df["n_algorithms_selected"] == c)
        ].copy()
        cdf = cdf[cdf["pre_selector"] != "BeamSearch_w5"]
        cdf = cdf[cdf["pre_selector"] != "GeneticAlgorithm_large"]
        cdf = cdf[cdf["pre_selector"] != "RandomLocalSearch_r5"]
        cdf["pre_selector"] = cdf["pre_selector"].replace(display_names)
        if cdf.shape[0] == 0:
            print("    Not enough data")
            continue
        cdf.sort_values(by=["pre_selector", "n_algorithms_selected"], inplace=True)
        fig = px.box(
            cdf,
            x="pre_selector",
            y="normalized_score",
            color="pre_selector",
            template="plotly_white",
        )
        fig.update_yaxes(matches=None, showticklabels=True)
        fig.update_layout(
            showlegend=False,
            xaxis_title=None,
            yaxis_title="Closed Gap",
            width=250,
            height=250,
            margin=dict(l=0, r=0, t=0, b=0),
            font=dict(size=9),
        yaxis=dict(
            tickmode="array",
            # tickvals=list(range(len(pivot_display.index))),
            # ticktext=pivot_display.index.tolist(),
            # autorange="reversed",
            # tickangle=30,
            tickfont=dict(size=9),
        ),
        xaxis=dict(
            tickmode="array",
            # tickvals=[str(c) for c in pivot_display.columns],
            # ticktext=[str(c) for c in pivot_display.columns],
            tickfont=dict(size=9),
            title_font=dict(size=9),
        ),
        )
        fig.write_image("/home/shavit/asf/paper/ol_v1/figures/pre_selection/pre_selector_" + scenario + f"_n{c}.pdf")
```

```python
pivot = (
    df.groupby(["pre_selector", "n_algorithms_selected"])["normalized_score"]
    .agg("mean")
    .unstack()
)
# pivot = pivot[pivot.index != "BruteForce"]
pivot = pivot[pivot.index != "BeamSearch_w5"]
pivot = pivot[pivot.index != "GeneticAlgorithm_large"]
pivot = pivot[pivot.index != "RandomLocalSearch_r5"]

pivot = pivot * 100  # Convert to percentage

# Rename index for display
pivot_display = pivot.rename(index=display_names)

import plotly.graph_objects as go

# Create text array with empty strings for NaN values
text_array = [
    [f"{v:.1f}" if not np.isnan(v) else "" for v in row] for row in pivot_display.values
]

# Target ~300x300 for the plot area, add space for labels/ticks/titles
plot_size = 225
left_margin = 140
bottom_margin = 60

fig = go.Figure(
    data=go.Heatmap(
        z=pivot_display.values,
        x=[str(c) for c in pivot_display.columns],
        y=pivot_display.index.tolist(),
        text=text_array,
        texttemplate="%{text}",
        textfont=dict(size=8),  # Small font to fit in cells
        colorscale="Plasma",
        zmin=np.nanmin(pivot_display.values),
        zmax=np.nanmax(pivot_display.values),
        hoverongaps=False,
        showscale=False,
        xgap=0,
        ygap=0,
    )
)

fig.update_layout(
    xaxis_title="Number of Selected Algorithms",
    yaxis_title=None,
    width=plot_size + left_margin,
    height=plot_size + bottom_margin,
    font=dict(size=8),
    yaxis=dict(
        tickmode="array",
        tickvals=list(range(len(pivot_display.index))),
        ticktext=pivot_display.index.tolist(),
        autorange="reversed",
        tickangle=30,
        tickfont=dict(size=8),
    ),
    xaxis=dict(
        tickmode="array",
        tickvals=[str(c) for c in pivot_display.columns],
        ticktext=[str(c) for c in pivot_display.columns],
        tickfont=dict(size=8),
        title_font=dict(size=8),
    ),
    margin=dict(l=5, r=5, t=5, b=5),
    plot_bgcolor="white",
)
fig.update_traces(showlegend=False)
fig.write_image("/home/shavit/asf/paper/ol_v1/figures/pre_selection/pre_selector_heatmap.pdf")
```

```python
pivot = (
    df.groupby(["pre_selector", "n_algorithms_selected"])["cpu_time_seconds"]
    .agg("mean")
    .unstack()
)
# pivot = pivot[pivot.index != "BruteForce"]
pivot = pivot[pivot.index != "BeamSearch_w5"]
pivot = pivot[pivot.index != "GeneticAlgorithm_large"]
pivot = pivot[pivot.index != "RandomLocalSearch_r5"]

# Don't multiply by 100 - this is time in seconds, not percentage

# Rename index for display
pivot_display = pivot.rename(index=display_names)

import plotly.graph_objects as go

# Create text array with appropriate formatting based on value magnitude
def format_time(v):
    if np.isnan(v):
        return ""
    elif v >= 1000:
        return f"{v/1000:.1f}k"  # Show as "1.2k" for thousands
    elif v >= 100:
        return f"{v:.0f}"  # No decimals for 100+
    elif v >= 10:
        return f"{v:.1f}"  # 1 decimal for 10-99
    elif v >= 1:
        return f"{v:.1f}"
    else:
        return f"{v:.2f}"  # 2 decimals for small values

text_array = [
    [format_time(v) for v in row] for row in pivot_display.values
]

# Target ~300x300 for the plot area, add space for labels/ticks/titles
plot_size = 225
left_margin = 140
bottom_margin = 60

# Use log scale for better color distribution since times vary widely
import numpy as np
log_values = np.log10(pivot_display.values + 0.01)  # Add small value to handle zeros

fig = go.Figure(
    data=go.Heatmap(
        z=log_values,  # Use log-transformed values for color
        x=[str(c) for c in pivot_display.columns],
        y=pivot_display.index.tolist(),
        text=text_array,  # But show original values as text
        texttemplate="%{text}",
        textfont=dict(size=7),  # Slightly smaller font to fit
        colorscale="Plasma",  # Better colorscale for time data
        zmin=np.nanmin(log_values),
        zmax=np.nanmax(log_values),
        hoverongaps=False,
        showscale=False,
        xgap=0,
        ygap=0,
    )
)

fig.update_layout(
    xaxis_title="Number of Selected Algorithms",
    yaxis_title=None,
    width=plot_size + left_margin,
    height=plot_size + bottom_margin,
    font=dict(size=8),
    yaxis=dict(
        tickmode="array",
        tickvals=list(range(len(pivot_display.index))),
        ticktext=pivot_display.index.tolist(),
        autorange="reversed",
        tickangle=30,
        tickfont=dict(size=8),
    ),
    xaxis=dict(
        tickmode="array",
        tickvals=[str(c) for c in pivot_display.columns],
        ticktext=[str(c) for c in pivot_display.columns],
        tickfont=dict(size=8),
        title_font=dict(size=8),
    ),
    margin=dict(l=5, r=5, t=5, b=5),
    plot_bgcolor="white",
)
fig.update_traces(showlegend=False)
fig.write_image("/home/shavit/asf/paper/ol_v1/figures/pre_selection/pre_selector_heatmap_time.pdf")
```
